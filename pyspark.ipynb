## Differen between dropDuplicates() vs distinct()
**distinct() -> removes duplicate rows (entire row match)**

**dropDuplicates([subset]) -> removes duplicate based on one or more specific columns**


```python
data = [
    (1, "Alice", 23),
    (2, "Bob", 34),
    (3, "Charlie", 29),
    (1, "Alice", 23),   # duplicate row
    (2, "Bob", 34),     # duplicate row
    (6, "Alice", 30)    # same name, different age
]

columns = ["id", "name", "age"]

df = spark.createDataFrame(data, columns)
display(df)
```

```python
display(df.distinct())
```

```python
display(df.dropDuplicates(["age"]))
```

### **What is groupBy() in PySpark?**

**The groupBy() transformation groups rows that have the same values in specified column(s).**

**After grouping, you must apply aggregation functions (like count, sum, avg, min, max).**

**Similar to GROUP BY in SQL.**

```python
data = [
    (1, "Alice", "NY", 2000),
    (2, "Bob", "CA", 1500),
    (3, "Charlie", "NY", 3000),
    (4, "David", "CA", 2500),
    (5, "Eve", "TX", 1800),
    (6, "Frank", "TX", 2200),
]

columns = ["id", "name", "state", "salary"]

df = spark.createDataFrame(data, columns)
display(df)
```

```python
display(df.groupBy('state').count())
```

```python
display(df.groupBy('state').agg({'salary': 'avg'}))
```

```python
display(df.groupBy('state').agg({'salary': 'sum'}))
```

```python
display(df.groupBy('state').agg({'salary': 'max'}))
```

```python
display(df.groupBy('state').agg({'salary': 'min'}))
```

## **What is a UDF?**

**A User Defined Function (UDF) lets you use custom Python logic inside Spark SQL/DataFrames.**

**You register your Python function as a UDF so Spark can distribute it across the cluster.**

**Use only if built-in Spark SQL functions can’t solve your problem (since UDFs can be slower).**


```python
data = [
    ("Alice", "Delhi", 10),
    ("Bob", "Mumbai", 25),
    ("Charlie", "Bangalore", 60),
    ("David", "Pune", None)
]

columns = ["name", "city", "age"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python

def age_grp(age):
    if age is None:
        return "UNKNOWN"
    elif age < 18:
        return "MINOR"
    elif age >= 18 and age < 40:
        return "ADULT"
    else:
        return "SENIOR"
age_grp(10)
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

age_grp_category= udf(age_grp,StringType())
```

```python
df.withColumn('age_cat',age_grp_category(df.age)).show()
```

## **partitionBy AND repartition in PySpark**

```python
from pyspark.sql import SparkSession, functions as F

spark = SparkSession.builder.getOrCreate()

# ---------- 1) Build a larger DataFrame (no RDD usage) ----------
categories = ["Electronics","Grocery","Clothing","Home","Sports"]
regions    = ["North","South","East","West","Central"]
years      = [2023, 2024, 2025]
months     = list(range(1, 13))

df_cat   = spark.createDataFrame([(c,) for c in categories], ["category"])
df_reg   = spark.createDataFrame([(r,) for r in regions],    ["region"])
df_year  = spark.createDataFrame([(y,) for y in years],      ["year"])
df_month = spark.createDataFrame([(m,) for m in months],     ["month"])
```

```python
grid = df_cat.crossJoin(df_reg).crossJoin(df_year).crossJoin(df_month)
```

```python
df = (
    grid
    .withColumn("n", F.floor(F.rand(seed=42) * 16 + 5))  # 5..20
    .select(
        "*",
        F.explode(F.sequence(F.lit(1), F.col("n"))).alias("txn_id")
    )
    .drop("n")
    .withColumn("sales", (F.rand(seed=7) * 900 + 100).cast("int"))   # 100..1000
    .withColumn("units", (F.rand(seed=11) * 9 + 1).cast("int"))      # 1..10
    .withColumn(
        "dt",
        F.to_date(F.concat_ws("-", F.col("year"), F.col("month"), F.lit(1)))
    )
)

print("Row count:", df.count())
display(df)
```

```python
df.write.mode('overwrite').partitionBy('region').parquet("/Volumes/development/data/files/sales")
```

```python
df.repartition(3).write.mode("overwrite").partitionBy("region").parquet("/Volumes/development/data/files/sales2")
```

```python
df.coalesce(1).write.mode('overwrite').partitionBy("region").parquet("/Volumes/development/data/files/sales2/")
```

**What is explode?**

In PySpark, explode() is used to transform an array or map column into multiple rows.

```python
data = [
    (1, ["apple", "banana", "orange"]),
    (2, ["grapes", "melon"]),
    (3, [])
]
df = spark.createDataFrame(data, ["id", "fruits"])
display(df)
```

```python
from pyspark.sql.functions import *
df.select("id",explode("fruits")).show()
```

```python
from pyspark.sql.functions import *
df.select("id",posexplode("fruits")).show()
```
